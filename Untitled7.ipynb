```python
%%writefile lorenz_sim/src/solver.py
import numpy as np
from abc import ABC, abstractmethod
from typing import Tuple, List

# 1. Arayüz: Sisteme yüklenecek dinamik denklemlerin uyması gereken şablon
class DynamicSystem(ABC):
    @abstractmethod
    def get_derivatives(self, state: np.ndarray) -> np.ndarray:
        """Sistemin türev vektörünü (dx/dt, dy/dt, dz/dt) döner."""
        pass

# 2. Somut Sistem: Lorenz Denklemlerinin Matematiksel Modeli
class LorenzSystem(DynamicSystem):
    def __init__(self, sigma: float = 10.0, rho: float = 28.0, beta: float = 8.0/3.0):
        self.sigma = sigma
        self.rho = rho
        self.beta = beta

    def get_derivatives(self, state: np.ndarray) -> np.ndarray:
        x, y, z = state[0], state[1], state[2]
        dx = self.sigma * (y - x)
        dy = x * (self.rho - z) - y
        dz = x * y - self.beta * z
        return np.array([dx, dy, dz], dtype=float)

# 3. Sayısal Çözücü Motoru (RK4 Integrator)
class RK4Solver:
    def __init__(self, system: DynamicSystem, dt: float = 0.01):
        self.system = system
        self.dt = dt

    def step(self, current_state: np.ndarray) -> np.ndarray:
        """Runge-Kutta 4. Dereceden İntegrasyon Adımı"""
        dt = self.dt

        k1 = self.system.get_derivatives(current_state)
        k2 = self.system.get_derivatives(current_state + 0.5 * dt * k1)
        k3 = self.system.get_derivatives(current_state + 0.5 * dt * k2)
        k4 = self.system.get_derivatives(current_state + dt * k3)

        next_state = current_state + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
        return next_state

    def simulate(self, initial_state: np.ndarray, num_steps: int) -> np.ndarray:
        """Belirtilen adım sayısı kadar simülasyonu koşturur ve tüm yörüngeyi döner."""
        trajectory = np.zeros((num_steps, 3), dtype=float)
        current_state = np.array(initial_state, dtype=float)

        for i in range(num_steps):
            trajectory[i] = current_state
            current_state = self.step(current_state)

        return trajectory
```

```python
%%writefile lorenz_sim/src/analytics.py
import numpy as np

class ChaosAnalytics:
    """İki yörünge arasındaki kaotik sapmayı ve diverjansı analiz eder."""

    @staticmethod
    def calculate_divergence(trajectory1: np.ndarray, trajectory2: np.ndarray) -> np.ndarray:
        """Her zaman adımında iki yörünge arasındaki Öklid mesafesini hesaplar."""
        if trajectory1.shape != trajectory2.shape:
            raise ValueError("Yörünge matrislerinin boyutları eşit olmalıdır!")

        # Vektörize Öklid uzaklığı hesaplama (Döngüsüz, yüksek performans)
        diff = trajectory1 - trajectory2
        distance = np.sqrt(np.sum(diff**2, axis=1))
        return distance
```

```python
%%writefile lorenz_sim/tests/test_simulation.py
import numpy as np
import pytest
from lorenz_sim.src.solver import LorenzSystem, RK4Solver
from lorenz_sim.src.analytics import ChaosAnalytics

def test_lorenz_equilibrium():
    # Lorenz sisteminin bilinen sabit noktalarından biri (0,0,0) kökenidir.
    # Başlangıç noktası 0 ise türevler 0 olmalı ve sistem hareket etmemelidir.
    system = LorenzSystem()
    initial_state = np.array([0.0, 0.0, 0.0])
    solver = RK4Solver(system, dt=0.01)

    next_state = solver.step(initial_state)
    assert np.allclose(next_state, initial_state)

def test_chaos_analytics_distance():
    # Analitik motorun Öklid mesafesini doğru hesaplayıp hesaplamadığını doğrular.
    traj1 = np.array([[0.0, 0.0, 0.0], [1.0, 1.0, 1.0]])
    traj2 = np.array([[3.0, 4.0, 0.0], [1.0, 1.0, 1.0]])

    distances = ChaosAnalytics.calculate_divergence(traj1, traj2)

    # İlk adımda 3-4-5 üçgeninden mesafe 5.0 çıkmalı, ikinci adımda çakıştıkları için 0.0 çıkmalı
    assert np.isclose(distances[0], 5.0)
    assert np.isclose(distances[1], 0.0)
```

```python
%%writefile app.py
import streamlit as st
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from lorenz_sim.src.solver import LorenzSystem, RK4Solver
from lorenz_sim.src.analytics import ChaosAnalytics

st.set_page_config(page_title="İTÜ CSE - Kaos Simülatörü", layout="wide")

st.title("🦋 Lorenz Çekicisi ve Kelebek Etkisi Simülatörü")
st.markdown("Bu interaktif yazılım, kaotik dinamik sistemleri **RK4 sayısal entegrasyonu** ile çözer ve başlangıç koşullarına hassasiyeti analiz eder.")

# Yan Panel Parametreleri
st.sidebar.header("🎛️ Sistem Parametreleri")
sigma = st.sidebar.slider("Sigma (Prandtl Sayısı)", 1.0, 20.0, 10.0)
rho = st.sidebar.slider("Rho (Rayleigh Sayısı - Kaos için >24.74)", 10.0, 40.0, 28.0)
beta = st.sidebar.slider("Beta (Geometrik Boyut)", 1.0, 5.0, 2.666)

st.sidebar.header("🚀 Simülasyon Ayarları")
num_steps = st.sidebar.slider("Adım Sayısı (Zaman Uzunluğu)", 500, 5000, 2000)
dt = st.sidebar.slider("Zaman Adımı (dt)", 0.001, 0.05, 0.01)

st.sidebar.header("🦋 Kelebek Etkisi (Başlangıç Koşulları)")
x1 = st.sidebar.number_input("1. Yörünge Başlangıç X", value=1.0)
delta_x = st.sidebar.number_input("2. Yörünge İçin Mikro Değişim (Delta X)", value=0.0001, format="%.5f")

# Hesaplamaları Başlatma
if st.sidebar.button("⚙️ Simülasyonu Çalıştır"):

    # 1. Sistemleri Kurma
    system = LorenzSystem(sigma=sigma, rho=rho, beta=beta)
    solver = RK4Solver(system, dt=dt)

    init_state1 = np.array([x1, 1.0, 1.0])
    init_state2 = np.array([x1 + delta_x, 1.0, 1.0])

    # 2. Sayısal Çözüm (RK4 Simülasyonu)
    traj1 = solver.simulate(init_state1, num_steps)
    traj2 = solver.simulate(init_state2, num_steps)

    # 3. Analitik Sapma Hesabı
    divergence = ChaosAnalytics.calculate_divergence(traj1, traj2)

    # Arayüz Raporlama
    col1, col2 = st.columns(2)

    with col1:
        st.subheader("🪐 3D Uzayda Kaotik Atraktör Yörüngesi")
        fig = plt.figure(figsize=(7, 6))
        ax = fig.add_subplot(111, projection='3d')

        # 1. Kelebeğin Rotası (Mavi)
        ax.plot(traj1[:, 0], traj1[:, 1], traj1[:, 2], color='teal', alpha=0.7, label="1. Yörünge", linewidth=1)
        # 2. Kelebeğin Rotası (Kırmızı)
        ax.plot(traj2[:, 0], traj2[:, 1], traj2[:, 2], color='crimson', alpha=0.6, label="2. Yörünge", linewidth=1, linestyle='--')

        ax.set_xlabel("X Ekseni")
        ax.set_ylabel("Y Ekseni")
        ax.set_zlabel("Z Ekseni")
        ax.legend()
        st.pyplot(fig)

    with col2:
        st.subheader("📉 Zaman İçinde İki Yörüngenin Kopuş Analizi")
        fig2, ax2 = plt.subplots(figsize=(7, 5))
        time_axis = np.arange(num_steps) * dt

        ax2.plot(time_axis, divergence, color='purple', label="Uzaklık (Öklid)")
        ax2.set_xlabel("Zaman (saniye)")
        ax2.set_ylabel("Yörüngeler Arası Mesafe")
        ax2.set_title("Kelebek Etkisinin Matematiksel İspatı")
        ax2.grid(True, linestyle=':')
        ax2.legend()
        st.pyplot(fig2)

    st.info(f"💡 **Hesaplamalı Bilim Notu:** İlk başta neredeyse tamamen çakışık olan (Fark: {delta_x}) iki sistem, kaotik bölgede ilerledikçe belirli bir saniyeden sonra üstel (exponential) olarak birbirinden kopmuştur. Bu durum deterministik sistemlerde uzun vadeli tahmin yapılamayacağının matematiksel kanıtıdır.")
```
